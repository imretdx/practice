```csharp
using Microsoft.DotNet.Interactive.Formatting;
Formatter.Register(typeof(ScottPlot.Plot), (p, w) => w.Write(((ScottPlot.Plot)p).GetPngHtml(600, 400)), HtmlFormatter.MimeType);
```

```csharp
// Данные замеров (ось X - Время в мс, ось Y - Количество потоков)
double[] times = new double[] { 120.0, 75.0, 42.0, 28.0, 31.0, 35.0 };
double[] threads = new double[] { 1, 2, 4, 8, 12, 16 };

ScottPlot.Plot plt = new();
plt.Title("Зависимость времени вычисления от количества потоков");
plt.XLabel("Время выполнения Solve (мс)");
plt.YLabel("Количество потоков");

plt.Add.Scatter(times, threads);
plt
```
